# Prompt Engineering

### Objective

The objective of this section is to improve the quality of the responses generated by the Large Language Model (LLM) through prompt engineering techniques. Instead of changing the model or providing external knowledge, carefully designed prompts are used to guide the model toward producing more accurate, professional, and reliable responses.

The responses generated using different prompt designs will be compared to evaluate their impact on relevance, clarity, and groundedness. This demonstrates how prompt engineering can improve the performance of an LLM even before introducing Retrieval-Augmented Generation (RAG).

## Prompt 1: Baseline Prompt

The baseline prompt consists of only the user's question without any additional instructions or context. This serves as the reference point for evaluating how prompt engineering techniques improve the quality of responses.

**Objective:**

* Observe the default behavior of the LLM.
* Establish a baseline for comparison with improved prompts.

```python
def generate_baseline_response(question):
    """
    Generates a response using only the user's question
    without any prompt engineering.
    """

    llm = ChatGoogleGenerativeAI(
        model="gemini-3.5-flash",
        temperature=0.2,
        google_api_key=os.environ["GOOGLE_API_KEY"]
    )

    response = llm.invoke(question)

    return response.text
```

```python
#Applying baseline prompt engineering
questions = [
    "What are the effects on the benefits I receive if my probation is extended?",
    "There has been a demise in my family last night, and I need to attend the last rites. How should I inform the office, and will I be granted leave?",
    "What should I do if I notice suspected harassment with my female colleague?"
]

print("BASELINE PROMPT RESPONSES")


for i, question in enumerate(questions, start=1):
    print(f"\nQuestion {i}")
    print("-" * 120)
    print(question)

    response = generate_baseline_response(question)

    print("\nResponse:\n")
    print(response)
```

The baseline prompt generated detailed and relevant responses for all three business questions. However, the responses were based entirely on the model's general knowledge and were not grounded in the organization's HR handbook. This led the model to make assumptions about company policies and include information that may not be applicable, such as references to U.S. employment laws and generic HR practices.

```python
#Role-based Prompt

def generate_role_based_response(question):
    """
    Generates a response using a role-based prompt.
    """

    prompt = f"""
You are an experienced Human Resources (HR) Assistant.

Your responsibilities are:
- Answer HR-related questions in a professional and concise manner.
- If the question requires company-specific policies that are not available, clearly mention that your response is based on general HR practices.
- Do not assume or fabricate organization-specific policies.
- Recommend referring to the official HR handbook or HR department whenever organization-specific information is required.

Employee Question:
{question}

Answer:
"""

    llm = ChatGoogleGenerativeAI(
        model="gemini-3.5-flash",
        temperature=0.2,
        google_api_key=os.environ["GOOGLE_API_KEY"]
    )

    response = llm.invoke(prompt)

    return response.text
```

```python
questions = [
    "What are the effects on the benefits I receive if my probation is extended?",
    "There has been a demise in my family last night, and I need to attend the last rites. How should I inform the office, and will I be granted leave?",
    "What should I do if I notice suspected harassment with my female colleague?"
]

print("ROLE-BASED PROMPT RESPONSES")


for i, question in enumerate(questions, start=1):
    print(f"\nQuestion {i}")
    print("-" * 120)
    print(question)

    response = generate_role_based_response(question)

    print("\nResponse:\n")
    print(response)
```

The role-based prompt produced responses that were more focused, professional, and transparent than those generated by the baseline prompt. Unlike the baseline, the model clearly stated that its answers were based on general HR practices and avoided making organization-specific assumptions. Although the responses were still not grounded in the company's HR handbook, the prompt successfully reduced hallucinations and encouraged the model to recommend consulting the official HR handbook or HR department whenever company-specific information was required.

```python
#Contextual and Structured Prompt

def generate_structured_response(question):
    """
    Generates a response using a contextual and structured prompt.
    """

    prompt = f"""
You are an experienced Human Resources (HR) Assistant.

Your goal is to provide accurate, professional, and employee-friendly HR guidance.

Follow these instructions carefully:
1. Answer only using general HR knowledge.
2. Do NOT assume or invent company-specific policies.
3. If the answer depends on company policy, clearly state that the information is unavailable and recommend referring to the official HR handbook or HR department.
4. Organize your response using the following structure:
   - Summary
   - Guidance
   - Recommendation
5. Keep the response concise, clear, and professional.

Employee Question:
{question}

Response:
"""

    llm = ChatGoogleGenerativeAI(
        model="gemini-3.5-flash",
        temperature=0.2,
        google_api_key=os.environ["GOOGLE_API_KEY"]
    )

    response = llm.invoke(prompt)

    return response.text
```

```python
# Business Questions

questions = [
    "What are the effects on the benefits I receive if my probation is extended?",
    "There has been a demise in my family last night, and I need to attend the last rites. How should I inform the office, and will I be granted leave?",
    "What should I do if I notice suspected harassment with my female colleague?"
]

print("CONTEXTUAL AND STRUCTURED PROMPT RESPONSES")

for i, question in enumerate(questions, start=1):
    print(f"\nQuestion {i}")
    print("-" * 120)
    print(question)

    response = generate_structured_response(question)

    print("\nResponse:\n")
    print(response)
```

The contextual and structured prompt generated the most consistent and well-organized responses among the three prompting techniques. By enforcing a predefined response structure and explicit constraints, the model presented information in a clearer and more user-friendly manner while continuing to avoid unsupported assumptions about company policies. Although the responses still relied on general HR knowledge due to the absence of organizational context, they were more transparent, easier to follow, and better suited for an HR assistance application.

# Retrieval-Augmented Generation (RAG)

## Objective

Although prompt engineering improved the quality and consistency of the responses, the LLM was still unable to provide organization-specific answers because it had no access to the company's HR handbook. As a result, the responses continued to rely on general HR knowledge and could not accurately answer policy-related questions.

To address this limitation, Retrieval-Augmented Generation (RAG) is introduced. RAG combines information retrieval with text generation by first retrieving relevant sections from the HR handbook and then supplying that information to the LLM as context before generating a response. This enables the model to produce responses that are more accurate, context-aware, and grounded in the organization's official policies.

The RAG pipeline implemented in this project consists of the following stages:

1. Load the HR handbook.
2. Extract text from the document.
3. Split the text into manageable chunks.
4. Generate vector embeddings for each chunk.
5. Store the embeddings in a FAISS vector database.
6. Retrieve the most relevant chunks for a user query.
7. Generate a final response using Gemini with the retrieved context.

### Splitting the Document into Chunks

Large Language Models have a limited context window and cannot efficiently process an entire document at once. Therefore, the HR handbook is divided into smaller, overlapping text chunks using LangChain's RecursiveCharacterTextSplitter.

Chunking improves retrieval performance by allowing the retriever to search for the most relevant sections of the handbook instead of the complete document. An overlap is maintained between consecutive chunks to preserve contextual continuity and reduce the chances of losing important information at chunk boundaries.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Initialize the text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

# Split the document into chunks
chunks = text_splitter.split_documents(documents)

print(f"Total chunks created: {len(chunks)}")
```

The HR handbook was successfully divided into 27 overlapping text chunks using the `RecursiveCharacterTextSplitter`. The selected chunk size and overlap preserved the context within each section while ensuring that the document was broken into smaller units suitable for semantic retrieval. These chunks will serve as the fundamental knowledge units for generating embeddings and retrieving relevant information during the RAG process.

```python
#Load Embedding Model

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

all-MiniLM-L6-v2 was selected because it is lightweight, computationally efficient, and widely used for semantic search and RAG applications. It generates high-quality sentence embeddings while maintaining fast inference, making it well suited for an HR handbook knowledge base.

```python
# Create the FAISS vector database

vector_store = FAISS.from_documents(
    documents=chunks,
    embedding=embedding_model
)

print("FAISS vector database created successfully.")
print(f"Total vectors stored: {vector_store.index.ntotal}")
```

The text chunks were successfully converted into vector embeddings using the Hugging Face embedding model and stored in a FAISS vector database. A total of 27 vectors were created, matching the number of document chunks. This confirms that each chunk has been successfully indexed and is now ready for semantic similarity search during the retrieval process.

### Creating a Retriever

A retriever is configured to perform semantic similarity searches. When a user submits a query, the retriever compares the query embedding with the stored document embeddings and returns the most relevant chunks from the HR handbook. These retrieved chunks will later be provided to the language model as context for generating grounded responses.

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

print("Retriever created successfully.")
```

```python
# Testing the retriever

query = "What happens if my probation period is extended?"

retrieved_docs = retriever.invoke(query)

print(f"Retrieved {len(retrieved_docs)} document chunks.\n")

for i, doc in enumerate(retrieved_docs, start=1):
    print(f"Retrieved Chunk {i}")
    print(doc.page_content[:600])
    print()
```

The retriever successfully performed a semantic search on the FAISS vector database and returned the three most relevant document chunks for the given query.

The retrieved content contained information directly related to the probation policy, including performance reviews, contract renewal eligibility, and the impact of probation extensions. This confirms that the vector database and retriever are functioning correctly and can retrieve relevant organizational knowledge based on semantic similarity.

## Building RAG pipeline

the retriever is integrated with the Gemini language model to create a RAG pipeline. When a user submits a query, the retriever first searches the FAISS vector database and retrieves the most relevant sections from the HR handbook. These retrieved sections are then supplied as context to the language model, enabling it to generate responses that are grounded in the organization's official policies rather than relying solely on its pretrained knowledge.

```python
prompt = ChatPromptTemplate.from_template("""
You are an HR Assistant.

Your task is to answer employee questions ONLY using the information provided in the HR handbook context below.

Instructions:
- Use only the retrieved context to answer the question.
- Do NOT add assumptions or information that is not present in the context.
- If the answer is not available in the context, reply:
  "I could not find this information in the HR handbook."
- Begin with a direct answer.
- Briefly explain the relevant policy in simple, professional language.
- Use bullet points whenever multiple policy details are involved.
- Keep the response concise and employee-friendly.

HR Handbook Context:
{context}

Employee Question:
{question}

Answer:
""")
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0.2,
    google_api_key=os.environ["GOOGLE_API_KEY"]
)
```

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)
```

```python
rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough(),
    }
    | prompt
    | llm
    | StrOutputParser()
)

print("RAG pipeline created successfully.")
```

```python
query = "What are the effects on the benefits I receive if my probation is extended?"

response = rag_chain.invoke(query)

print(response)
```

The RAG pipeline successfully retrieved relevant information from the HR handbook and generated a response grounded in the organization's policies. Unlike the prompt engineering approaches, the response contained specific policy details regarding probation extensions, including restrictions on annual leave encashment, internal role transfers, performance bonuses, and seniority accrual. This demonstrates that integrating document retrieval with the language model significantly improves the accuracy and relevance of responses for organization-specific queries.

## Final Project Section

```python
# Create a RAG prompt that instructs Gemini to answer
# using only the information retrieved from the HR handbook.

rag_prompt = """
You are an HR Assistant for the organization.

Answer the employee's question using ONLY the information
provided in the retrieved HR handbook context.

If the answer cannot be found in the provided context,
clearly state that the information is not available in the
retrieved HR handbook context. Do not make assumptions or
invent company policies.

Keep the response professional, clear, and concise.

Retrieved HR Handbook Context:
{context}

Employee Question:
{question}

Answer:
"""

print("RAG prompt template created successfully.")
```

```python
def generate_rag_response(question):
    """
    Retrieves relevant HR handbook chunks and generates
    an answer using Gemini based only on the retrieved context.
    """

    # Retrieve relevant document chunks
    retrieved_docs = retriever.invoke(question)

    # Combine retrieved chunks into a single context
    context = "\n\n".join(
        doc.page_content for doc in retrieved_docs
    )

    # Fill the RAG prompt with retrieved context and question
    formatted_prompt = rag_prompt.format(
        context=context,
        question=question
    )

    # Generate response using the LLM
    response = llm.invoke(formatted_prompt)

    # Return both the generated response and retrieved documents
    return response.text, retrieved_docs


print("RAG response generation function created successfully.")
```

The RAG response-generation function was created successfully. It will retrieve relevant handbook chunks, pass them as context to Gemini, and return both the generated answer and retrieved documents. This gives us the basis for evaluating whether the final answers are properly grounded in the HR handbook.

```python
#Baseline RAG - test questions
questions = [
    "What are the effects on the benefits I receive if my probation is extended?",

    "There has been a demise in my family last night, and I need to attend the last rites. How should I inform the office, and will I be granted leave?",

    "What should I do if I notice suspected harassment with my female colleague?"
]

rag_results = {}

for i, question in enumerate(questions, start=1):
    answer, retrieved_docs = generate_rag_response(question)

    rag_results[f"Question {i}"] = {
        "question": question,
        "answer": answer,
        "retrieved_docs": retrieved_docs
    }

    print("=" * 80)
    print(f"QUESTION {i}")
    print("=" * 80)
    print(question)

    print("\nRAG RESPONSE:")
    print(answer)

    print("\nRETRIEVED CHUNKS:")
    for j, doc in enumerate(retrieved_docs, start=1):
        print(f"\n--- Retrieved Chunk {j} ---")
        print(doc.page_content[:1000])

    print("\n")

print("Baseline RAG testing completed successfully.")
```

The baseline RAG system worked successfully for all three questions. The retrieved chunks were relevant to the respective queries, and the generated answers were grounded in the HR handbook rather than relying only on general HR knowledge. The system also correctly avoided inventing a specific bereavement duration when that information was not present in the retrieved context.

```python
def evaluate_rag_response(question, answer, retrieved_docs):
    """
    Provides a simple qualitative evaluation of a RAG response
    based on the retrieved HR handbook context.
    """

    context = "\n".join(
        doc.page_content for doc in retrieved_docs
    )

    evaluation_prompt = f"""
You are evaluating an HR question-answering system.

Evaluate the answer against the retrieved HR handbook context.

Question:
{question}

Retrieved HR Handbook Context:
{context}

Generated Answer:
{answer}

Evaluate the response on these two criteria:

1. Groundedness:
Does the answer rely only on information supported by the retrieved
HR handbook context? Identify any unsupported claims or assumptions.

2. Relevance:
Does the answer directly address the employee's question using the
most relevant information from the retrieved context?

Give a score from 1 to 5 for each criterion and provide a brief
explanation.

Return the evaluation in this format:

Groundedness Score: X/5
Groundedness Comment: ...

Relevance Score: X/5
Relevance Comment: ...
"""

    evaluation = llm.invoke(evaluation_prompt)

    return evaluation.text


rag_evaluations = {}

for question_number, result in rag_results.items():

    evaluation = evaluate_rag_response(
        result["question"],
        result["answer"],
        result["retrieved_docs"]
    )

    rag_evaluations[question_number] = evaluation

    print("=" * 80)
    print(question_number)
    print("=" * 80)
    print(evaluation)
    print()

print("Baseline RAG evaluation completed successfully.")
```

The baseline RAG evaluation shows strong performance across all three questions. Each response received 5/5 for groundedness and 5/5 for relevance, indicating that the answers were supported by the retrieved HR handbook content and directly addressed the employee queries. This establishes a strong baseline for comparison with the fine-tuned RAG configurations.

Next we will test at least 5 different RAG hyperparameter combinations systematically and evaluate each configuration on the same three questions.

```python
rag_configs = [
    {
        "config": "Configuration 1 - Baseline",
        "chunk_size": 1000,
        "chunk_overlap": 200,
        "search_type": "similarity",
        "k": 3
    },
    {
        "config": "Configuration 2",
        "chunk_size": 500,
        "chunk_overlap": 100,
        "search_type": "similarity",
        "k": 3
    },
    {
        "config": "Configuration 3",
        "chunk_size": 750,
        "chunk_overlap": 150,
        "search_type": "similarity",
        "k": 3
    },
    {
        "config": "Configuration 4",
        "chunk_size": 1000,
        "chunk_overlap": 200,
        "search_type": "similarity",
        "k": 5
    },
    {
        "config": "Configuration 5",
        "chunk_size": 1000,
        "chunk_overlap": 200,
        "search_type": "mmr",
        "k": 3
    }
]

print("RAG fine-tuning configurations defined successfully.")
print(f"Total configurations to test: {len(rag_configs)}")

for config in rag_configs:
    print(
        f"{config['config']}: "
        f"chunk_size={config['chunk_size']}, "
        f"overlap={config['chunk_overlap']}, "
        f"search={config['search_type']}, "
        f"k={config['k']}"
    )
```

We have defined five controlled RAG configurations that vary document chunking and retrieval parameters. The baseline configuration represents the current system, while the remaining configurations test different combinations to determine whether retrieval quality and answer performance can be improved.

```python
# Store the vector database and retriever for each configuration
rag_systems = {}

# We use the documents loaded from the HR handbook earlier
# and rebuild the chunks for each configuration.

for config in rag_configs:

    # Create text splitter using the configuration parameters
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=config["chunk_size"],
        chunk_overlap=config["chunk_overlap"]
    )

    # Split the original HR handbook documents
    config_chunks = text_splitter.split_documents(documents)

    # Create FAISS vector database
    config_vector_db = FAISS.from_documents(
        config_chunks,
        embedding_model
    )

    # Create retriever using the selected search method and k
    config_retriever = config_vector_db.as_retriever(
        search_type=config["search_type"],
        search_kwargs={"k": config["k"]}
    )

    # Store everything for later evaluation
    rag_systems[config["config"]] = {
        "config": config,
        "chunks": config_chunks,
        "vector_db": config_vector_db,
        "retriever": config_retriever
    }

    print(
        f"{config['config']} built successfully "
        f"({len(config_chunks)} chunks)."
    )

print("\nAll RAG configurations built successfully.")
```

All five RAG configurations were built successfully. The different chunking settings produced different numbers of document chunks, while Configurations 4 and 5 retained the original 27-chunk structure and changed the retrieval behavior.

```python
# Initialize storage for all fine-tuning experiment results
fine_tuning_results = {}

print("fine_tuning_results initialized successfully.")
```

```python
def generate_config_rag_response(question, config_retriever):
    """
    Generate a RAG response using a specific retriever configuration.
    """

    # Retrieve relevant chunks
    retrieved_docs = config_retriever.invoke(question)

    # Combine retrieved content
    context = "\n\n".join(
        doc.page_content for doc in retrieved_docs
    )

    # Format the RAG prompt
    formatted_prompt = rag_prompt.format(
        context=context,
        question=question
    )

    # Generate answer
    response = llm.invoke(formatted_prompt)

    return response.text, retrieved_docs


print("Configuration-specific RAG response function created successfully.")
```

```python
config_name = "Configuration 1 - Baseline"
system = rag_systems[config_name]

fine_tuning_results[config_name] = {}

for i, question in enumerate(questions, start=1):

    answer, retrieved_docs = generate_config_rag_response(
        question,
        system["retriever"]
    )

    fine_tuning_results[config_name][f"Question {i}"] = {
        "question": question,
        "answer": answer,
        "retrieved_docs": retrieved_docs
    }

    print("=" * 100)
    print(f"{config_name} — QUESTION {i}")
    print("=" * 100)
    print(question)

    print("\nRAG RESPONSE:")
    print(answer)

    print("\nRETRIEVED CHUNKS:")
    for j, doc in enumerate(retrieved_docs, start=1):
        print(f"\n--- Chunk {j} ---")
        print(doc.page_content[:500])

    print()

print(f"{config_name} testing completed successfully.")
```

Configuration 1 reproduced the baseline RAG behavior successfully. All three questions received responses grounded in relevant handbook sections, with the system correctly identifying when specific information was not available in the retrieved context. The retrieved chunks were also appropriately related to each question.

## Evaluation Methodology

Each response was manually evaluated on a 1–5 scale using the generated answer, the original question, and—in the case of RAG—the retrieved context.

Groundedness: Measures whether the answer is supported by the provided/retrieved source context.

5 — Fully supported
4 — Mostly supported with minor unsupported details
3 — Partially supported
2 — Mostly unsupported
1 — Unsupported

Relevance: Measures how directly and completely the answer addresses the user's question.

5 — Direct and complete answer
4 — Mostly answers the question
3 — Partially answers the question
2 — Limited relevance
1 — Does not adequately answer the question

```python
import pandas as pd

comparison_data = []

for config_name, questions_data in evaluation_results.items():

    groundedness_scores = [
        q["Groundedness"]
        for q in questions_data.values()
    ]

    relevance_scores = [
        q["Relevance"]
        for q in questions_data.values()
    ]

    avg_groundedness = sum(groundedness_scores) / len(groundedness_scores)
    avg_relevance = sum(relevance_scores) / len(relevance_scores)

    overall_score = (avg_groundedness + avg_relevance) / 2

    comparison_data.append({
        "Configuration": config_name,
        "Avg Groundedness": round(avg_groundedness, 2),
        "Avg Relevance": round(avg_relevance, 2),
        "Overall Score": round(overall_score, 2)
    })

comparison_df = pd.DataFrame(comparison_data)

comparison_df = comparison_df.sort_values(
    by="Overall Score",
    ascending=False
).reset_index(drop=True)

print("RAG Fine-Tuning Performance Comparison:\n")
print(comparison_df.to_string(index=False))
```
